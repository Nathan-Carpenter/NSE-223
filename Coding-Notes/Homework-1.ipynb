The rest mass of a baseball is 145g and the speed of light, $c = 2.99792458 × 10^8 m/s$. What is the mass of a baseball when thrown at $v= 169.1$ km/h (the fastest recorded pitch)? How fast does it have to move to have a mass of 1.45 kg?

### **Chapter 1 Programming Project 1**
The cubic equation
$$x^3+4b^2x=2c^3$$
has a root given by
$$x = d-\frac{b^2}{d}$$
where
$$d^3 = c^3 + \sqrt{b^6+c^6}$$
This method was developed by Thomas Harriot (1560-1621) who also introduced the less than and greater than symbols. Write a program that prompts the user for coefficients of a general cubic,
$$Ax^3+Bx^2+Cx+D$$
and determines if the cubic can be solved via Harriot's method (i.e., $B=0$). If it can be solved via Harriot's method, then print the solution. Also print the residual from the root, that is the value you get when you plug each into the original equation. Make sure that you do not divide by zero, and that your method can handle imaginary roots.

```python
import math

A = float(input("Enter coefficient A: "))
B = float(input("Enter coefficient B: "))
C = float(input("Enter coefficient C: "))
D = float(input("Enter coefficient D: "))

if B != 0:
    print("This cubic cannot be solved by Harriot's method (B must be 0).")
else:
    # Avoid division by zero
    if A == 0:
        print("Not a cubic equation (A cannot be 0).")
    else:
        # x^3 + (C/A)x + (D/A) = 0
        # Harriot's method is for x^3 + 4b^2 x = 2c^3

        b2 = C / (4*A)
        c3 = -D / (2*A)

        # d^3 = c^3 + sqrt(b^6 + c^6)
        d_cubed = c3 + math.sqrt(b2**3 + c3**2)
        d = d_cubed**(1/3) 

        # Avoid division by zero
        if d == 0:
            print("d is zero, cannot compute root (division by zero).")
        else:
            x_root = d - b2 / d
            print("Root by Harriot's method: x =",x_root)
            # Compute residual
            residual = A*x_root**3 + C*x_root + D
            print("Residual (plug root into original equation):",residual)
```

### **Problem 2.2**
If we list all the natural numbers below 10 that are multiples of 3 or 7, we get 3, 6, 7, and 9. The sum of these multiples is 25. Find the sum of all multiples of 3 or 7 below 10000.

```python
below = 10000 # Set the upper limit
natural_numbers = [] # List to store the natural numbers
for i in range(below+1):
    if i != 0 and (i % 3 == 0 or i % 7 == 0): # Check to see if i is multiple of 3 or 7
        natural_numbers.append(i)

#print(natural_numbers) # Print the list of natural numbers, to check
print("Sum:", sum(natural_numbers))
```

### **Chapter 2 Programming Project 1** - Nuclear Reaction Q Values
Write a Python code that asks the user to input the masses (in amu) of two reactants in a nuclear reaction and the two products of the reaction. The code will output the Q value of the reaction in MeV. Here is an example reaction:
$$\binom{A_x}{Z_x}X+\binom{A_x}{Z_x}x → \binom{A_y}{Z_y}Y+\binom{A_y}{Z_y}y$$
The user should be able to enter zero for the mass in case the reaction is a decay (i.e., has only one reactant) or is a reaction that has only one product. Use $1 amu = 931.494061 MeV/c^2$

```python
reactant_1_amu = float(input("Enter the atomic mass of reactant 1 (in amu): "))
reactant_2_amu = float(input("Enter the atomic mass of reactant 2 (in amu): "))
product_1_amu = float(input("Enter the atomic mass of product 1 (in amu): "))
product_2_amu = float(input("Enter the atomic mass of product 2 (in amu): "))

# Q = [Mass of reactants - Mass of products] * c^2
c2 = 931.494061 # MeV/amu

Q_value = (reactant_1_amu + reactant_2_amu - product_1_amu - product_2_amu) * c2
print("The Q-value of the reaction is:", Q_value, "MeV")
```

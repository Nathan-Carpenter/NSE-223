```python
import numpy as np

import matplotlib.pyplot as plt

# Constants
Lambda = 5e-5  # mean generation time (s)
beta = np.array([0.00021, 0.00142, 0.00128, 0.00257, 0.00075, 0.00027])
beta_bar = 0.0065
t_half = np.array([56, 23, 6.2, 2.3, 0.61, 0.23])  # half-lives (s)
lambda_i = np.log(2) / t_half

# Inhour equation RHS as a function of s
def inhour_rhs(s):
    return s * (Lambda + np.sum(beta / (s + lambda_i)))

# Reactivity values to plot
rho_values = [-1, 0, 0.1 * beta_bar, beta_bar]
rho_labels = [r'$\rho=-1$', r'$\rho=0$', r'$\rho=0.1\bar{\beta}$', r'$\rho=\bar{\beta}$']

# s range (avoid singularities)
s_min = -2.5
s_max = 0.05
s = np.linspace(s_min, s_max, 2000)
for lmbd in lambda_i:
    s = s[np.abs(s + lmbd) > 0.01]  # avoid singularities

# Plot
plt.figure(figsize=(10, 6))
plt.plot(s, inhour_rhs(s), label='Inhour RHS')
for rho, label in zip(rho_values, rho_labels):
    plt.axhline(y=rho, linestyle='--', label=label)

plt.ylim(s_min, s_max)
plt.xlabel('s (1/s)')
plt.ylabel('RHS of Inhour Equation')
plt.title('Graphical Solution of the Inhour Equation')
plt.legend()
plt.grid(True)
plt.show()
```

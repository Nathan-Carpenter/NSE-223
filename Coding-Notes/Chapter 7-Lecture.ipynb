# Chapter 7 - Gaussian Elimination

So much of what we end up doing when we solve problems in science and engineering is taking governing ordinary or partial differential equations that describe the physics, chemistry and even biology and turning them into systems of linear, or possibly nonlinear, algebraic equations.  We need to know how to efficiently and accurately solve these linear systems of equations.  [Nonlinear equations are often solved by iterating through approximate linear systems of equations...]

Let's look at a simple example - just math.  No physics.  A three by three linear system of equations:

$$ 3 x_1 + 2 x_2 + x_3 = 6, $$
$$ -x_1 + 4 x + 5 x_3 = 8, $$
$$ 2 x_1 - 8x_2 + 10 x_3 = 4. $$

In matrix form, this is 

$$\begin{pmatrix}
3 & 2 & 1 \\
-1 & 4 & 5 \\
2 &  -8 & 10
\end{pmatrix} 
\begin{pmatrix}
x_1 \\
x_2 \\
x_3 
\end{pmatrix} =
\begin{pmatrix}
6 \\
8 \\
4 
\end{pmatrix}$$

Or, in a shorthand called *augmented matrix form*, it looks like this:

$$\begin{pmatrix}
    3 & 2 & 1 &\bigm| & 6 \\
    -1 & 4 & 5 &\bigm| & 8 \\
    2 & -8 & 10 &\bigm| & 4 
\end{pmatrix}$$

Let's store this 3 by 4 matrix for use later...

```python
import numpy as np
aug_matrix = np.matrix([(3.0,2,1,6),(-1,4,5,8),(2,-8,10,4)])
print(aug_matrix)
```

We know how to manually manipulate this system to solve it, by using the tools of elementary algebra to eliminate variables by adding and subtracting equations from each other.  

For instance, we we take the second row, and add 1/3 times the first row, we can replace the second row (and eliminate $x_1$ from the second equation)...

### Gaussian Elimination (Step-by-Step)

Start with the augmented matrix:

$$
\left[\begin{array}{ccc|c}
3 & 2 & 1 & 6 \\
-1 & 4 & 5 & 8 \\
2 & -8 & 10 & 4
\end{array}\right]
$$

---

### Step 1: Eliminate entries below the first pivot $$a_{11}=3$$

Row operations:
$$
\begin{aligned}
R_2 &\leftarrow R_2 + \tfrac{1}{3}R_1 \\
R_3 &\leftarrow R_3 - \tfrac{2}{3}R_1
\end{aligned}
$$

Resulting matrix:

$$
\left[\begin{array}{ccc|c}
3 & 2 & 1 & 6 \\
0 & \tfrac{14}{3} & \tfrac{16}{3} & 10 \\
0 & -\tfrac{28}{3} & \tfrac{28}{3} & 0
\end{array}\right]
$$

```python
#add row 2 to 1/3 times row 1
row13 = aug_matrix[0]/3 #row1 * 1/3
new_row2 = aug_matrix[1] + row13 #add 1/3 row 1 to row 2
#replace row 2
aug_matrix[1,:] = new_row2
print("New matrix =\n",aug_matrix)
```

This eliminated $x_1$ from the second equation.   Next step:  eliminate $x_1$ from the third equation by adding -2/3 times row 1 to row 3:

```python
#add row 3 to -2/3 times row 1
row23 = -2*aug_matrix[0]/3 #row1 * -2/3
new_row3 = aug_matrix[2] + row23 #add -2/3 row 1 to row 3
#replace row 3
aug_matrix[2] = new_row3
print("New matrix =\n",aug_matrix)
```

Now, eliminate $x_2$ from the third equation by adding to row 3 the quantity $\frac{(9 \frac{1}{3})}{(4 \frac{2}{3})}$ times row 1 to row3...

$$
\left[\begin{array}{ccc|c}
3 & 2 & 1 & 6 \\
0 & \tfrac{14}{3} & \tfrac{16}{3} & 10 \\
0 & -\tfrac{28}{3} & \tfrac{28}{3} & 0
\end{array}\right]
$$

---

### Step 2: Eliminate entry below the second pivot


$$
a_{22} = \tfrac{14}{3}
$$

Row operation:
$$
R_3 \leftarrow R_3 + 2R_2
$$

Resulting matrix:

$$
\left[\begin{array}{ccc|c}
3 & 2 & 1 & 6 \\
0 & \tfrac{14}{3} & \tfrac{16}{3} & 10 \\
0 & 0 & 20 & 20
\end{array}\right]
$$

```python
#add row 3 to 9.33333/4.66666 times row 2
modrow = (9+1./3)/(4+2./3)*aug_matrix[1] #row2 * (-9+1./3)/(4+2.0/3)
new_row3 = aug_matrix[2] + modrow #add -2/3 row 1 to row 3
#replace row 3
aug_matrix[2] = new_row3.copy()
print("New matrix =\n",aug_matrix)
```

Our original system now has been transformed into this equivalent system:

$$\begin{pmatrix}
    3 & 2 & 1 &\bigm| & 6 \\
    0 & 4 \frac{2}{3} & 5 \frac{1}{3} &\bigm| & 10 \\
    0 & 0 & 20 &\bigm| & 20 
\end{pmatrix}$$

We can easily solve this system using *back substitution* - starting with the last row, solve each equation in succession.  In this example, solve the third equation for $x_3$, plug this into the second equation and solve for $x_2$, and then substitute $x_3$ and $x_2$ into the first equation to solve for $x_1$.

### Back Substitution

Third row:
$$
20z = 20 \Rightarrow z = 1
$$

Second row:
$$
\tfrac{14}{3}y + \tfrac{16}{3}(1) = 10
\Rightarrow 14y + 16 = 30
\Rightarrow y = 1
$$

First row:
$$
3x + 2(1) + 1 = 6
\Rightarrow x = 1
$$

---

```python
#backsubstitution
x3 = aug_matrix[2,3]/aug_matrix[2,2] #solve for x3
print("x3 =",x3)
#now solve for x2
x2 = (aug_matrix[1,3] - x3*aug_matrix[1,2])/aug_matrix[1,1]
print("x2 =",x2)
#now solve for x1
x1 = (aug_matrix[0,3] - x3*aug_matrix[0,2]-
x2*aug_matrix[0,1])/aug_matrix[0,0]
print("x1 =",x1)
```

Let's check our answer but computing the residual $r = A x - b$:

```python
A = np.matrix([(3.0,2,1),(-1,4,5),(2,-8,10)])
x = np.array([1,1,1])
b = np.array([6,8,4])
print(np.dot(A,x),"-",b,"=",np.dot(A,x)-b)
```

Sweet.  We did it.  This is an illustration of vanilla, no bells or whistles, Gaussian Elimination.  We can write the back substitution part very cleanly and abstractly using the following formula:

$$ x_i = \frac{1}{A_{ii}} \left( b_i - \sum\limits_{j=i+1}^{3} A_{ij} x_j \right), \qquad i=1, 2, 3$$

Here we can see that this follows our general approach when we solved for $x_2$ and $x_1$. The solution for some $x_i$ is the RHS constant, minus the solutions from variables farther in the vector, and then all of that is divided by the value on the diagonal. 

## Let's make a function for solving 3x3 systems using Gaussian Elimination...

The procedure we used to solve the system given in Eq. (7.1) can be generalized to any
coefficients and righthand side. We could just copy and paste the code above, and then change
the numbers if we wanted to solve another system, but that would be prone to errors. In
the next code block, we define a function that mimics our algorithm above to change the
augmented matrix into a form ready for back substitution. The function is then tested on the
system in Eq. (7.1).


$$\begin{pmatrix}
3 & 2 & 1 \\
-1 & 4 & 5 \\
2 &  -8 & 10
\end{pmatrix} 
\begin{pmatrix}
x_1 \\
x_2 \\
x_3 
\end{pmatrix} =
\begin{pmatrix}
6 \\
8 \\
4 
\end{pmatrix}$$

\begin{aligned}
a_{00}x_0 + a_{01}x_1 + a_{02}x_2 &= b_0 \\
a_{10}x_0 + a_{11}x_1 + a_{12}x_2 &= b_1 \\
a_{20}x_0 + a_{21}x_1 + a_{22}x_2 &= b_2
\end{aligned}

```python
import numpy as np

def GaussElim33(A,b):
    """create a Gaussian elimination matrix for a 3x3 system
    Args:
        A: 3 by 3 array
        b: array of length 3
    Returns:
        augmented matrix ready for back substitution
    """
    #create augmented matrix
    aug_matrix = np.zeros((3,4))
    aug_matrix[0:3,0:3] = A
    aug_matrix[:,3] = b
    #augmented matrix is created

    #for column in [0, 1, 2]:
    for column in range(0,3):
        for row in range(column+1,3):
            mod_row = aug_matrix[row,:]
            mod_row = mod_row - (mod_row[column]/aug_matrix[column,column]*
                        aug_matrix[column,:])
            aug_matrix[row] = mod_row
    return aug_matrix


A = np.matrix([(3,2,1),(-1,4,5),(2,-8,10)])
b = np.array([6,8,4])

#test function on the problem above
aug = GaussElim33(A,b)
print(aug)
```

And now, the back substitution...


$$ x_i = \frac{1}{A_{ii}} \left( b_i - \sum\limits_{j=i+1}^{2} A_{ij} x_j \right), \qquad i=2, 1, 0$$

```python
def BackSub33(aug_matrix,x):
    """back substitute a 3 by 3 system after Gaussian elimination
    Args:
        aug_matrix: augmented matrix with zeros below the diagonal
        x: length 3 vector to hold solution
    Returns:
        nothing
    Side Effect:
        x now contains solution
    """
    #start at the end
    for row in [2,1,0]:   # row <-> i
        RHS = aug_matrix[row,3]
        for column in range(row+1,3):  # column <-> j
            RHS -= x[column]*aug_matrix[row,column]
            # Pause here to illustrate the importance of proper indention...
        x[row] = RHS/aug_matrix[row,row]
    return

x = np.zeros(3)
BackSub33(aug,x)
print("The solution is ", x)
```

## Gaussian Elimination for a General System

Would be nice to have something that worked no matter how big the linear algebra problem was, right?  I mean, do you really want to have to rewrite these functions if you have bigger or smaller matrices and vectors?  No, you don't.  Trust me.

The main change will be replace the number 3 with a variable $N$, such that our $A$ matrix is $N \times N$ and the right hand side vector $b$ is of length $N$.

```python
import numpy as np

def GaussElim(A,b):
    """create a Gaussian elimination matrix for an NxN system
    Args:
        A: N by N array
        b: array of length N
    Returns:
        augmented matrix ready for back substitution
    """
    #create augmented matrix
    [Nrow, Ncol] = A.shape
    assert Nrow == Ncol
    N = Nrow
    assert b.size == N
    aug_matrix = np.zeros((N,N+1))
    aug_matrix[0:N,0:N] = A
    aug_matrix[:,N] = b
    #augmented matrix is created
    for column in range(0,N):
        for row in range(column+1,N):
            mod_row = aug_matrix[row,:]
            mod_row -= (mod_row[column]/aug_matrix[column,column]*
                        aug_matrix[column,:])
            aug_matrix[row] = mod_row
    return aug_matrix

#let’s try it on a 4 x 4 to start
A = np.array([(3.0,2,1,1),(-1,4,5,-2),(2,-8,10,-3),(2,3,4,5)])
answer = np.arange(4)+1.0 #1,2,3,4
b = np.dot(A,answer)
aug = GaussElim(A,b)
print(aug)
```

Now, let's generalize the backsubstitution function...

```python
def BackSub(aug_matrix,x):
    """back substitute a N by N system after Gaussian elimination
    Args:
        aug_matrix: augmented matrix with zeros below the diagonal
        x: length N vector to hold solution
    Returns:
        nothing
    Side Effect:
        x now contains solution
    """
    N = x.size
    [Nrow,Ncol] = aug_matrix.shape
    assert Nrow+1 == Ncol
    assert N == Nrow
    #start at the end
    for row in range(N-1,-1,-1):
        RHS = aug_matrix[row,N]
        for column in range(row+1,N):
            RHS -= x[column]*aug_matrix[row,column]
            # Pause here to illustrate the importance of proper indention...
        x[row] = RHS/aug_matrix[row,row]
    return

x = np.zeros(4)
BackSub(aug,x)
print("The solution is ", x)
```

Sweet!  We have a generic solver.  Let's do something with it!  Let's fill a large matrix with 2.01 on the diagonals, and -1 on the immediate off diagonals, and zero everywhere else, and then make a simple right hand side.

Here you can see how to quickly fill in a matrix. We'll do a few things:
* fill up diagonal first by using the dimensions of the matrix to get the indexes
* Fill up off-diagonals similarly, but adjust the index for above and below the diagonal


```python
import matplotlib.pyplot as plt
import numpy as np

mat_size = 100
A = np.zeros((mat_size,mat_size))
b = np.zeros(mat_size)

#indices of the diagonal elements, length = mat_size
diag = np.arange(mat_size)
A[diag,diag] = 2.01

# from 1 to mat_size-1; the row index; column = row-1 
belowDiagRow = np.arange(1,mat_size)
A[belowDiagRow,belowDiagRow-1] = -1

# from 0 to mat_size-2; the row index; column = row+1 
aboveDiagRow = np.arange(mat_size-1)
A[aboveDiagRow,aboveDiagRow+1] = -1

print(A)
```

## Round off and pivoting...

When matrices have very large differences in the magnitudes of the elements, Gaussian Elimination can behave more poorly, the reason for which is the fact that digital computers use arithmetic with finite precision.  Large numbers being divided by small numbers can lead to significant round-off error, which can propagate through all the calculations in the algorithm.  

First let's create a matrix that have similar numbers except one (note that we see numpy automatically format the values of the other numbers, like we talked about a few weeks ago)... 

```python
epsilon = 1e-14
A = np.array([(epsilon,-1,1),(-1,2,-1),(2,-1,0)])
print(A)
```

Now choose something pretty simple for the RHS:

```python
b = np.array([0,0,1.0])
print(b)
```

```python
aug_mat = GaussElim(A,b)
x = np.zeros(3)
BackSub(aug_mat,x)
print("The solution is",x)
print("The residual is",b-np.dot(A,x))
```

Here we compute the residual by calculating b using our solution for x (np.dot(A,x)), and then subtract that from the actual value of b that we defined. If we have an exact solution, the residual should be zero. 

This residual is pretty large.  Part of the reason for this is that we have not used the largest row element as the pivot element, or the element we divide by when doing the elimination. In this example, the pivot element is near zero and finite precision error is magnified when we divide by this small number.



Suppose the small number has a tiny rounding error:
$$
\frac{1}{0.0001} = 10000
$$
but if the true value were slightly different,
$$
\frac{1}{0.0001001} \approx 9990
$$

To correct this issue we could rearrange the rows so that we divide by larger elements. This
is called *row pivoting*.

Here is another illustration of the need for pivoting...

```python
aug_mat = GaussElim(A,b)
x = np.zeros(2)
BackSub(aug_mat,x)
print(x)
```

Now, what if we switch the rows in A and try the same thing:

```python
A = np.array([(0,1.0),(1.0,0)])
b = np.array([6.,2.0])
print("A =\n",A)
print("b =",b)
aug_mat = GaussElim(A,b)
x = np.zeros(2)
BackSub(aug_mat,x)
print("x =",x)
```

We naively divided by zero, since we didn't have the rows arranged to avoid that.  

To implement pivoting, we need a function that will __swap rows__ in our augmented matrix:

```python
def swap_rows(A, a, b):
    """Swap two rows in a matrix: switch row a with row b
    args:
        A: matrix to perform row swaps on
        a: row index of matrix
        b: row index of matrix
    returns: nothing
    side effects:
        changes A to have rows a and b swapped
    """
    assert (a>=0) and (b>=0)
    N = A.shape[0] #number of rows
    print(N,a,b)
    assert (a<N) and (b<N) #less than because 0-based indexing
    temp = A[a,:].copy()
    A[a,:] = A[b,:].copy()
    A[b,:] = temp.copy()
    return
    
print("Before swap, A =\n",A)
swap_rows(A,0,1)
print("After swapping 0 and 1, A =\n",A)
```

The next step is figuring out how to swap during each step. At the beginning of the solve
we want find the maximum row element magnitude for each row and store it as a vector.

```python
N = 5
A = np.random.rand(N,N)
print("A =\n",A)
s = np.zeros(N)
count = 0
for row in A:
    s[count] = np.max(np.fabs(row))
    count += 1
print("s =",s)
```

Then we will have to figure out which row has the largest <u> scaled </u> element in the pivot
position and call the swap_rows function. We will use the argmax function which returns
the index of the largest element in a vector.

- New function - argmax: returns the index of the largest element of a vector.  


Scaled partial pivoting selects the pivot by comparing each candidate pivot to the scale of its row.

For each row i, define the row scale factor
$$s_i = \max_j |a_{ij}|$$

At elimination step k, choose the pivot row p such that
$$\frac{|a_{p k}|}{s_p}
=
\max_{i \ge k}
\frac{|a_{i k}|}{s_i}
$$
Then swap row k with row p.



### Why "relative size" matters for choosing a pivot (scaled pivoting)

If one row is $(10,11,10)$ and another is $(2,1,1)$, the pivot value **2** can actually be a *better* choice than **10**.

**Why?**

- In the first row, the largest entry is about $11$, so $10$ is not especially dominant.
- In the second row, $2$ is much larger than the other entries in that row ($1$ and $1$), so it is a stronger pivot *relative to that row’s scale*.

This motivates **scaled partial pivoting**.

Instead of selecting a pivot based only on $|a_{i1}|$, we compare the pivot to the scale of its row:

$$
\frac{|a_{i1}|}{\max\left(|\text{row}_i|\right)}
$$

This helps avoid choosing pivots that look large in absolute value but are not strong compared to the other entries in the same row.



```python
pivot_column = 2
largest_pos = np.argmax(np.fabs(A[:,pivot_column]/s)) #division is element-wise

print("A =\n",A)
print("s =",s)

print("Largest scaled element in column",
      pivot_column,"is in row",largest_pos)
```

Now can put this all together in a new version of Gaussian elimination. This function will
find the largest elements in each row of the matrix, then proceed with Gaussian elimination
where the pivot element is the element that is the largest element remaining in the
row.

```python
def swap_rows(A, a, b):
    """Rows two rows in a matrix, switch row a with row b
    
    args:
        A: matrix to perform row swaps on
        a: row index of matrix
        b: row index of matrix
        
    returns: nothing
    
    side effects:
    changes A to rows a and b swapped
    """
    assert (a>=0) and (b>=0)
    N = A.shape[0] #number of rows
    assert (a<N) and (b<N) #less than because 0-based indexing
    temp = A[a,:].copy()
    A[a,:] = A[b,:].copy()
    A[b,:] = temp.copy()
    
def BackSub(aug_matrix,x):
    """back substitute a N by N system after Gaussian elimination
    
    Args:
        aug_matrix: augmented matrix with zeros below the diagonal
        x: length N vector to hold solution
    Returns:
        nothing
    Side Effect:
    x now contains solution
    """
    N = x.size
    for row in range(N-1,-1,-1):
        RHS = aug_matrix[row,N]
        for column in range(row+1,N):
            RHS -= x[column]*aug_matrix[row,column]
        x[row] = RHS/aug_matrix[row,row]
    return
```

```python
def GaussElimPivotSolve(A,b,LOUD=0):
    """create a Gaussian elimination with pivoting matrix for a system
    
    Args:
        A: N by N array
        b: array of length N
    Returns:
        solution vector in the original order
    """
    import numpy as np
    [Nrow, Ncol] = A.shape
    assert Nrow == Ncol
    N = Nrow
    #create augmented matrix
    aug_matrix = np.zeros((N,N+1))
    aug_matrix[0:N,0:N] = A
    aug_matrix[:,N] = b
    #augmented matrix is created
    
    #create scale factors
    s = np.zeros(N)
    count = 0
    for row in aug_matrix[:,0:N]: #don't include b
        s[count] = np.max(np.fabs(row))
        count += 1
    if LOUD:
        print("s =",s)
    if LOUD:
        print("Original Augmented Matrix is\n",aug_matrix)
    #perform elimination
    for column in range(0,N):
        # Find the row (at or below the current pivot row)
        # with the largest scaled pivot candidate in this column
        largest_pos = np.argmax(np.fabs(aug_matrix[column:N,column]/s[column])) + column #Does it make sense here?
        # largest_pos = np.argmax(np.fabs(aug_matrix[column:N,column]/s[column:N])) + column
```

```python
def GaussElimPivotSolve(A,b,LOUD=0):
    """create a Gaussian elimination with pivoting matrix for a system
    
    Args:
        A: N by N array
        b: array of length N
    Returns:
        solution vector in the original order
    """
    import numpy as np
    [Nrow, Ncol] = A.shape
    assert Nrow == Ncol
    N = Nrow
    #create augmented matrix
    aug_matrix = np.zeros((N,N+1))
    aug_matrix[0:N,0:N] = A
    aug_matrix[:,N] = b
    #augmented matrix is created
    
    #create scale factors
    s = np.zeros(N)
    count = 0
    for row in aug_matrix[:,0:N]: #don't include b
        s[count] = np.max(np.fabs(row))
        count += 1
    if LOUD:
        print("s =",s)
    if LOUD:
        print("Original Augmented Matrix is\n",aug_matrix)
    #perform elimination
    for column in range(0,N):
        # Find the row (at or below the current pivot row)
        # with the largest scaled pivot candidate in this column
        # largest_pos = np.argmax(np.fabs(aug_matrix[column:N,column]/s[column])) + column #Does it make sense here?
        largest_pos = np.argmax(np.fabs(aug_matrix[column:N,column]/s[column:N])) + column
                
        if (largest_pos != column):
            if (LOUD):
                print("Largest scaled element in column", column,"is in row",largest_pos)
                print("Swapping row",column,"with row",largest_pos)
                print("Pre swap\n",aug_matrix)
            swap_rows(aug_matrix,column,largest_pos)
            #re-order s
            tmp = s[column]
            s[column] = s[largest_pos]
            s[largest_pos] = tmp
            if (LOUD):
                print("A =\n",aug_matrix)
        #finish off the row
        for row in range(column+1,N):
            mod_row = aug_matrix[row,:]
            mod_row = mod_row - mod_row[column]/aug_matrix[column,column]*aug_matrix[column,:]
            aug_matrix[row] = mod_row
    #now back solve
    x = np.zeros(N)
    if LOUD:
        print("Final aug_matrix is\n",aug_matrix)
    BackSub(aug_matrix,x)
    return x
```

## Time to Solution for Gaussian Elimination

There is a detailed discussion in McClarren on the operation count for Gaussian Elimination, and you can read that and do the math if you choose. 

When we talk about operations that an algorithm is going to take, we use $O(n^z)$ notation. This notation gives us an idea of how a problem will scale. A system that will scales linearly will be defined as $O(n)$. 

The upshot, though, is that GE's computational cost scales as $O(N^3)$, where $N$ is the number of unknowns in the linear system.  Let's see if we can observe this trend by running a bunch of random tests...

We're using a the time library to test timing of a bunch of random tests. 


```python
import time
import matplotlib.pyplot as plt
num_tests = 12
N = 2**np.arange(num_tests)
times = np.zeros(num_tests)
for test in range(1,num_tests):
    A = np.random.rand(N[test],N[test])
    b = np.dot(A,np.ones(N[test]))
    x = np.zeros(N[test])
    start = time.time()
    aug = GaussElim(A,b)
    BackSub(aug,x)
    end = time.time()
    times[test] = end-start
plt.loglog(N,times,"o-", label="Gaussian Elimination")

y_comp = times.copy()/4
for comp_place in range(num_tests-1,0,-1):
    #because x goes up by factor 2 each time, time should go up by 8
    y_comp[comp_place -1] = np.exp(np.log(y_comp[test])-3*(
        np.log(N[test])-np.log(N[comp_place-1])))
    #t2/t1 = (n2/n1)^3 
plt.loglog(N[4:num_tests],y_comp[4:num_tests],"r",linewidth=4, label="$O(N^3)$")
#make a comparison line with slope 2

y_comp = times.copy()*4
for comp_place in range(num_tests-1,0,-1):
    y_comp[comp_place -1] = np.exp(np.log(y_comp[test])-2*(
        np.log(N[test])-np.log(N[comp_place-1])))
    #t2/t1 = (n2/n1)^2
plt.loglog(N[4:num_tests],y_comp[4:num_tests],"g:",linewidth=4, label="$O(N^2)$")
plt.legend("$O(N^2)$")

plt.xlabel("Number of Equations")
plt.ylabel("Seconds to complete solve")

plt.legend()
plt.tight_layout()

plt.show()
```

The solid red line is a slope 3 on a log-log scale (our theoretical prediction). The dotted line is a slope of 2.
